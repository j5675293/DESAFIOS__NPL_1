```python
# Se importa las librerias necesarias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import gensim
from gensim.models import Word2Vec
from sklearn.manifold import TSNE
import plotly.express as px
import plotly.graph_objects as go
from tqdm.notebook import tqdm
```

```python
# 1. Se carga y preprocesamiento de datos
print("Cargando datos...")
# Debe leer el archivo de canciones de los Beatles
try:
    # Lee el archivo como texto simple
    with open('songs_dataset/beatles.txt', 'r', encoding='utf-8') as file:
        lyrics = file.readlines()
except UnicodeDecodeError:
    # Intenta con otra codificación si falla la primera
    with open('songs_dataset/beatles.txt', 'r', encoding='latin1') as file:
        lyrics = file.readlines()

# Se crea DataFrame con las letras
df = pd.DataFrame(lyrics, columns=['lyrics'])

# Se limpia las líneas vacías y eliminar espacios en blanco al inicio/final
df['lyrics'] = df['lyrics'].str.strip()
df = df[df['lyrics'].str.len() > 0]

# Se muestra la información del dataset
print(f"Número total de canciones/documentos: {len(df)}")
print("\nPrimeras 5 canciones:")
print(df.head())
```

```python
# 2. Preprocesamiento de dataset
def preprocess_text(text):
    """
    Función para preprocesar el texto:
    - Convertir a minúsculas
    - Eliminar caracteres especiales
    - Tokenizar (usando método simple para evitar dependencias)
    - Eliminar stopwords
    - Lematizar palabras
    """
    if isinstance(text, str):
        # Se convierte a minúsculas
        text = text.lower()

        # Se elimina caracteres especiales y números
        text = re.sub(r'[^\w\s]', ' ', text)
        text = re.sub(r'\d+', ' ', text)

        # Se tokeniza de manera simple (sin usar nltk.tokenize)
        tokens = text.split() # Reemplacé word_tokenize de NLTK por problemas de iteracción

        # Se elimina stopwords (aseguramos que stopwords esté disponible)
        try:
            stop_words = set(stopwords.words('english'))
        except LookupError:
            # Si falla, crea un conjunto básico de stopwords en inglés
            stop_words = {'a', 'an', 'the', 'and', 'or', 'but', 'if', 'because', 'as', 'what',
                         'when', 'where', 'how', 'who', 'which', 'this', 'that', 'these', 'those',
                         'then', 'just', 'so', 'than', 'such', 'both', 'through', 'about', 'for',
                         'is', 'are', 'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had',
                         'having', 'do', 'does', 'did', 'doing', 'i', 'me', 'my', 'myself', 'we',
                         'our', 'ours', 'ourselves', 'you', "you're", "you've", "you'll", "you'd",
                         'your', 'yours', 'yourself', 'yourselves', 'he', 'him', 'his', 'himself',
                         'she', "she's", 'her', 'hers', 'herself', 'it', "it's", 'its', 'itself',
                         'they', 'them', 'their', 'theirs', 'themselves', 'to', 'from', 'up', 'down',
                         'in', 'out', 'on', 'off', 'over', 'under', 'again', 'further', 'then', 'once'}

        tokens = [word for word in tokens if word not in stop_words and len(word) > 2]

        # Se lematiza o simplemente usa las palabras como están si WordNetLemmatizer no está disponible
        try:
            lemmatizer = WordNetLemmatizer()
            tokens = [lemmatizer.lemmatize(word) for word in tokens]
        except LookupError:
            # Si no se puede lematizar, usar los tokens tal cual
            pass

        return tokens
    else:
        return []

print("\nPreprocesando textos...")
# Aplica preprocesamiento a todas las canciones
df['tokens'] = df['lyrics'].apply(preprocess_text)

# Elimina canciones que quedaron sin tokens después del preprocesamiento
df = df[df['tokens'].map(len) > 0]
print(f"Número de canciones después de eliminar documentos vacíos: {len(df)}")

# Ver cómo quedaron los tokens de las primeras canciones
print("\nEjemplo de tokenización (primeras 3 canciones):")
for i, tokens in enumerate(df['tokens'].head(3)):
    print(f"Canción {i+1}: {tokens[:15]}...")
```

# 2. Entrenamiento con Gensim
Se ejecuta lo siguiente:
*   Se utiliza Word2Vec de Gensim con arquitectura Skip-gram
*   Vectores de dimensión 100
*   Parámetros optimizados para corpus de canciones


```python
# 3. Entrenamiento de Word2Vec con Gensim
print("\nEntrenando modelo Word2Vec...")
# Parámetros del modelo
vector_size = 100  # Dimensionalidad de los vectores
window = 5         # Tamaño de la ventana contextual
min_count = 5      # Frecuencia mínima de palabras
workers = 4        # Número de hilos para entrenamiento paralelo
sg = 1             # 1 para skip-gram; 0 para CBOW

# Se entrena modelo
model = Word2Vec(
    df['tokens'].tolist(),
    vector_size=vector_size,
    window=window,
    min_count=min_count,
    workers=workers,
    sg=sg
)

# Información del modelo
print(f"Modelo entrenado con {len(model.wv.index_to_key)} palabras únicas")
```

```python
# 4. Exploración de los embeddings
print("\nAnálisis de palabras similares:")

# Palabras relevantes para analizar (ejemplos de palabras frecuentes en canciones de los Beatles)
palabras_clave = ['love', 'girl', 'day', 'know', 'time', 'good', 'night']

# Encuentra las palabras más similares para cada palabra clave
for palabra in palabras_clave:
    try:
        similares = model.wv.most_similar(palabra, topn=5)
        print(f"\nPalabras más similares a '{palabra}':")
        for word, score in similares:
            print(f"{word}: {score:.4f}")
    except KeyError:
        print(f"La palabra '{palabra}' no está en el vocabulario del modelo")
```

En el modelo a ejecutar se ha encontrando similitudes entre palabras con significados muy distintos. Por ejemplo, "love" y "four", "girl" y "money", "night" y "little". Esto sugiere que el corpus de entrenamiento puede estar influenciado por un contexto específico, como canciones, literatura, o algún conjunto de datos con asociaciones estilísticas particulares. Evaluaciones posteriores puede confirmar tal episodio.

```python
# 5. Visualización de embeddings con t-SNE y Plotly
print("\nPreparando visualización t-SNE...")

# Se selecciona las palabras más frecuentes para visualizar (para evitar saturar el gráfico)
top_words = 150

# Se obtiene las palabras más frecuentes basadas en el modelo
words = model.wv.index_to_key[:top_words]

# Se extra los vectores para estas palabras
word_vectors = np.array([model.wv[word] for word in words])

# Se reduce dimensionalidad con t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
word_vectors_2d = tsne.fit_transform(word_vectors)

# Se crea un DataFrame para plotly
df_plot = pd.DataFrame({
    'word': words,
    'x': word_vectors_2d[:, 0],
    'y': word_vectors_2d[:, 1]
})

# Se visualiza con Plotly
fig = px.scatter(
    df_plot, x='x', y='y', text='word',
    title=f'Word Embeddings de letras de The Beatles - Top {top_words} palabras',
    labels={'x': 't-SNE 1', 'y': 't-SNE 2'}
)

# Se personaliza gráfico
fig.update_traces(
    textposition='top center',
    marker=dict(size=8, opacity=0.8),
    textfont=dict(size=10, color='darkblue')
)

fig.update_layout(
    height=700,
    width=900,
    hovermode='closest',
    template='plotly_white'
)

# Se muestra gráfico
fig.show()
```

# 4. Análisis de agrupaciones de palabras

```python
# 6. Análisis de agrupaciones de palabras
print("\nAnalizando agrupaciones por categorías...")

# Se define categorías de palabras para analizar
categories = {
    'Emociones': ['love', 'feel', 'heart', 'cry', 'happy', 'sad'],
    'Tiempo': ['day', 'night', 'time', 'year', 'today', 'tomorrow'],
    'Personas': ['girl', 'boy', 'man', 'woman', 'friend', 'baby'],
}

# se crea visualización de clusters por categoría
fig_categories = go.Figure()

# Colores para cada categoría
colors = ['#FF5733', '#33BBFF', '#33FF57']

# Se agrega cada categoría al gráfico
for i, (category, cat_words) in enumerate(categories.items()):
    # Se filtra palabras que están en el vocabulario
    valid_words = [word for word in cat_words if word in model.wv]

    if valid_words:
        # Se obtiene vectores para estas palabras
        cat_vectors = np.array([model.wv[word] for word in valid_words])

        # Se aplica t-SNE a estos vectores
        cat_tsne = TSNE(n_components=2, random_state=42, perplexity=min(5, len(valid_words)-1) if len(valid_words) > 1 else 1)
        cat_vectors_2d = cat_tsne.fit_transform(cat_vectors)

        # Se agrega al gráfico
        for j, word in enumerate(valid_words):
            fig_categories.add_trace(
                go.Scatter(
                    x=[cat_vectors_2d[j, 0]],
                    y=[cat_vectors_2d[j, 1]],
                    mode='markers+text',
                    name=word,
                    text=word,
                    textposition='top center',
                    marker=dict(size=10, color=colors[i % len(colors)]),
                    legendgroup=category,
                    hoverinfo='text',
                    hovertext=f"{word} ({category})"
                )
            )

# Layout para el gráfico de categorías
fig_categories.update_layout(
    title='Agrupación de palabras por categorías semánticas en canciones de The Beatles',
    height=600,
    width=800,
    template='plotly_white',
    legend_title_text='Palabras',
    hovermode='closest'
)

# Se muestra gráfico de categorías
fig_categories.show()
```

Este análisis sugiere que en las letras de The Beatles, las palabras tienden a agruparse en torno a los siguientes temas semánticos principales:

- Amor y emociones (Positivas y Negativas),  un núcleo fuerte de palabras emocionales, con el amor y la felicidad como el centro más denso, y la tristeza/llanto como un subgrupo.
- Tiempo y temporalidad, una clara concentración de palabras relacionadas con el tiempo, indicando una posible preocupación por el paso del tiempo, momentos específicos, o el día a día.
- Relaciones personales, un grupo de palabras que denotan roles o relaciones entre personas, aunque estas parecen tener una conexión más difusa con los otros temas, lo que podría significar que estas relaciones se entrelazan con una variedad de emociones y temporalidades.

# 5. Análisis de conexiones semanticas entre palabras

```python
# 7. Análisis de conexiones semánticas entre palabras
print("\nAnalizando conexiones semánticas...")

# Palabras principales para analizar
central_words = ['love', 'girl', 'day']

# Se crea gráfico de conexiones
fig_connections = go.Figure()

# Para cada palabra central, mostrar sus palabras más similares
for central_word in central_words:
    if central_word in model.wv:
        # Se obtiene las palabras más similares
        similar_words = [word for word, _ in model.wv.most_similar(central_word, topn=7)]

        # Se obtiene vectores 2D para estas palabras
        words_to_plot = [central_word] + similar_words
        vectors = np.array([model.wv[word] for word in words_to_plot])

        # Se aplica t-SNE
        tsne_conn = TSNE(n_components=2, random_state=42, perplexity=min(5, len(words_to_plot)-1))
        vectors_2d = tsne_conn.fit_transform(vectors)

        # Se agrega nodo central
        fig_connections.add_trace(
            go.Scatter(
                x=[vectors_2d[0, 0]],
                y=[vectors_2d[0, 1]],
                mode='markers+text',
                marker=dict(size=15, color='red'),
                text=central_word,
                name=central_word,
                textposition='middle center',
                textfont=dict(size=12, color='white')
            )
        )

        # Se agrega nodos conectados
        fig_connections.add_trace(
            go.Scatter(
                x=vectors_2d[1:, 0],
                y=vectors_2d[1:, 1],
                mode='markers+text',
                marker=dict(size=10),
                text=similar_words,
                name=f'Relacionadas con {central_word}',
                textposition='top center'
            )
        )

        # Se agrega líneas de conexión
        for i in range(1, len(vectors_2d)):
            fig_connections.add_shape(
                type="line",
                x0=vectors_2d[0, 0],
                y0=vectors_2d[0, 1],
                x1=vectors_2d[i, 0],
                y1=vectors_2d[i, 1],
                line=dict(color="lightgrey", width=1)
            )

# Layout para el gráfico de conexiones
fig_connections.update_layout(
    title='Conexiones semánticas entre palabras clave en canciones de The Beatles',
    height=600,
    width=800,
    template='plotly_white',
    showlegend=True
)

# Se muestra gráfico de conexiones
fig_connections.show()
```

- Temas centrales,  "Love", "girl" y "day" son conceptos fundamentales en las canciones de The Beatles.
- Algunas palabras, como "money" y "lonely", tiende a ser conectores importantes, apareciendo en el ámbito semántico de múltiples palabras clave centrales.
- Los puntos con más líneas saliendo de ellos son más "centrales" en la red de conexiones.

# Entrenamiento con spaCy

#### Se utiliza spaCy, que ofrece un procesamiento lingüístico más avanzado y preciso que NLTK. Este código permite comparar los resultados entre ambos enfoques. El desarrollo de código tiene base estructural a NLTK.

```python
# Se importa las librerias requeridos
import spacy
from tqdm.notebook import tqdm
import gensim
from gensim.models import Word2Vec
from sklearn.manifold import TSNE
import plotly.express as px
import plotly.graph_objects as go
```

```python
# Se instala spaCy
!pip install -U spacy
!python -m spacy download en_core_web_sm
```

```python
# 1. Carga de datos
print("Cargando datos...")
try:
    # Leer el archivo como texto simple
    with open('songs_dataset/beatles.txt', 'r', encoding='utf-8') as file:
        lyrics = file.readlines()
except UnicodeDecodeError:
    # Intentar con otra codificación si falla la primera
    with open('songs_dataset/beatles.txt', 'r', encoding='latin1') as file:
        lyrics = file.readlines()

# Crear DataFrame con las letras
df = pd.DataFrame(lyrics, columns=['lyrics'])

# Limpiar las líneas vacías y eliminar espacios en blanco al inicio/final
df['lyrics'] = df['lyrics'].str.strip()
df = df[df['lyrics'].str.len() > 0]

# Mostrar la información del dataset
print(f"Número total de canciones/documentos: {len(df)}")
print("\nPrimeras 5 canciones:")
print(df.head())
```

# 1. Preprocesamiento

```python
# 2. Preprocesamiento con spaCy
print("\nCargando modelo de spaCy...")
nlp = spacy.load("en_core_web_sm")

def preprocess_with_spacy(text):
    """
    Función para preprocesar el texto usando spaCy:
    - Parseo con spaCy
    - Eliminación de stopwords, puntuación y números
    - Lematización
    """
    if isinstance(text, str):
        # se procesa el texto con spaCy
        doc = nlp(text.lower())

        # Se filtra tokens: eliminar stopwords, puntuación, espacios y números
        # También se eliminan palabras muy cortas (longitud < 3)
        tokens = [token.lemma_ for token in doc
                if not token.is_stop
                and not token.is_punct
                and not token.is_space
                and not token.like_num
                and len(token.text) > 2]

        return tokens
    else:
        return []

# Se procesa las letras en lotes pequeños para evitar problemas de memoria
print("\nPreprocesando textos con spaCy...")
batch_size = 100
num_batches = (len(df) + batch_size - 1) // batch_size

tokens_list = []
for i in tqdm(range(num_batches)):
    start_idx = i * batch_size
    end_idx = min((i + 1) * batch_size, len(df))
    batch = df['lyrics'][start_idx:end_idx]

    batch_tokens = [preprocess_with_spacy(text) for text in batch]
    tokens_list.extend(batch_tokens)

df['tokens'] = tokens_list

# Se elimina canciones que quedaron sin tokens después del preprocesamiento
df = df[df['tokens'].map(len) > 0]
print(f"Número de canciones después de eliminar documentos vacíos: {len(df)}")

# Se observa cómo quedaron los tokens de las primeras canciones
print("\nEjemplo de tokenización con spaCy (primeras 3 canciones):")
for i, tokens in enumerate(df['tokens'].head(3)):
    print(f"Canción {i+1}: {tokens[:15]}...")
```

```python
# 3. Entrenamiento de Word2Vec
print("\nEntrenando modelo Word2Vec...")
# Parámetros del modelo
vector_size = 100  # Dimensionalidad de los vectores
window = 5         # Tamaño de la ventana contextual
min_count = 5      # Frecuencia mínima de palabras
workers = 4        # Número de hilos para entrenamiento paralelo
sg = 1             # 1 para skip-gram; 0 para CBOW

# Entrenamiento de modelo
model = Word2Vec(
    df['tokens'].tolist(),
    vector_size=vector_size,
    window=window,
    min_count=min_count,
    workers=workers,
    sg=sg
)

# Información del modelo a entrenado
print(f"Modelo entrenado con {len(model.wv.index_to_key)} palabras únicas")
```

```python
# 4. Exploración embeddings
print("\nAnálisis de palabras similares:")

# Palabras relevantes para analizar
palabras_clave = ['love', 'girl', 'day', 'know', 'time', 'good', 'night']

# Encuentra las palabras más similares para cada palabra clave
for palabra in palabras_clave:
    try:
        similares = model.wv.most_similar(palabra, topn=5)
        print(f"\nPalabras más similares a '{palabra}':")
        for word, score in similares:
            print(f"{word}: {score:.4f}")
    except KeyError:
        print(f"La palabra '{palabra}' no está en el vocabulario del modelo")
```

- Este análisis de "palabras más similares" se basa en las incrustaciones de palabras (Word Embeddings), que representan el significado de las palabras en un espacio vectorial. La puntuación (cercana a 1.0) indica la similitud de coseno, es decir, qué tan cerca están los vectores de esas palabras en el espacio semántico.
- El análisis reafirma la importancia de temas como el amor, las relaciones humanas ("girl", "boy"), el tiempo ("day", "night", "time"), y las emociones ("lonely", "heart") en sus letras que inicialmente se habia descrito.

```python
# 5. Visualización de embeddings con t-SNE y Plotly
print("\nPreparando visualización t-SNE...")

# Se selecciona las palabras más frecuentes para visualizar
top_words = 150

# Se obtiene las palabras más frecuentes basadas en el modelo
words = model.wv.index_to_key[:top_words]

# Se extrae los vectores para estas palabras
word_vectors = np.array([model.wv[word] for word in words])

# Se reduce dimensionalidad con t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
word_vectors_2d = tsne.fit_transform(word_vectors)

# Se crea un DataFrame para plotly
df_plot = pd.DataFrame({
    'word': words,
    'x': word_vectors_2d[:, 0],
    'y': word_vectors_2d[:, 1]
})

# Se visualiza con Plotly
fig = px.scatter(
    df_plot, x='x', y='y', text='word',
    title=f'Word Embeddings (spaCy) de letras de The Beatles - Top {top_words} palabras',
    labels={'x': 't-SNE 1', 'y': 't-SNE 2'}
)

# Personalizacion de  gráfico
fig.update_traces(
    textposition='top center',
    marker=dict(size=8, opacity=0.8),
    textfont=dict(size=10, color='darkblue')
)

fig.update_layout(
    height=700,
    width=900,
    hovermode='closest',
    template='plotly_white'
)

# Se muestra grafico plotly
fig.show()
```

- El gráfico muestra un mapa semántico donde las palabras se agrupan en función de sus contextos de aparición y significado en las letras de The Beatles. Cada punto es una palabra.
- La disposición de las palabras valida la eficacia de los Word Embeddings y t-SNE para capturar relaciones semánticas. Las palabras con significados similares (ej. "love", "feel", "happy" o "sad", "lonely") están consistentemente agrupadas.

```python
# 6. Análisis de agrupaciones de palabras
print("\nAnalizando agrupaciones por categorías...")

# Se define categorías de palabras para analizar
categories = {
    'Emociones': ['love', 'feel', 'heart', 'cry', 'happy', 'sad'],
    'Tiempo': ['day', 'night', 'time', 'year', 'today', 'tomorrow'],
    'Personas': ['girl', 'boy', 'man', 'woman', 'friend', 'baby'],
}

# Se crea visualización de clusters por categoría
fig_categories = go.Figure()

# Colores para cada categoría
colors = ['#FF5733', '#33BBFF', '#33FF57']

# Se agrega cada categoría al gráfico
for i, (category, cat_words) in enumerate(categories.items()):
    # Se filtra palabras que están en el vocabulario
    valid_words = [word for word in cat_words if word in model.wv]

    if valid_words:
        # Se obtiene vectores para estas palabras
        cat_vectors = np.array([model.wv[word] for word in valid_words])

        # Se aplica t-SNE a estos vectores
        cat_tsne = TSNE(n_components=2, random_state=42, perplexity=min(5, len(valid_words)-1) if len(valid_words) > 1 else 1)
        cat_vectors_2d = cat_tsne.fit_transform(cat_vectors)

        # Se agrega al gráfico
        for j, word in enumerate(valid_words):
            fig_categories.add_trace(
                go.Scatter(
                    x=[cat_vectors_2d[j, 0]],
                    y=[cat_vectors_2d[j, 1]],
                    mode='markers+text',
                    name=word,
                    text=word,
                    textposition='top center',
                    marker=dict(size=10, color=colors[i % len(colors)]),
                    legendgroup=category,
                    hoverinfo='text',
                    hovertext=f"{word} ({category})"
                )
            )

# Layout para el gráfico de categorías
fig_categories.update_layout(
    title='Agrupación de palabras por categorías semánticas (spaCy) en canciones de The Beatles',
    height=600,
    width=800,
    template='plotly_white',
    legend_title_text='Palabras',
    hovermode='closest'
)

# Se muestra gráfico de categorías
fig_categories.show()
```

- Los puntos representan palabras, y la proximidad entre ellos indica similitud semántica. Los colores asignados a los puntos representan categorías semánticas (emociones, tiempo, personas). Sin embargo se evidencia que estan disperson, esto puede ser por numero de corpus.


```python
# 7. Análisis de conexiones semánticas entre palabras
print("\nAnalizando conexiones semánticas...")

# Palabras principales para analizar
central_words = ['love', 'girl', 'day']

# Se crea gráfico de conexiones
fig_connections = go.Figure()

# Para cada palabra central, se muestra sus palabras más similares
for central_word in central_words:
    if central_word in model.wv:
        # Obtiene las palabras más similares
        similar_words = [word for word, _ in model.wv.most_similar(central_word, topn=7)]

        # Obtiene vectores 2D para estas palabras
        words_to_plot = [central_word] + similar_words
        vectors = np.array([model.wv[word] for word in words_to_plot])

        # Aplicar t-SNE
        tsne_conn = TSNE(n_components=2, random_state=42, perplexity=min(5, len(words_to_plot)-1))
        vectors_2d = tsne_conn.fit_transform(vectors)

        # Agrega nodo central
        fig_connections.add_trace(
            go.Scatter(
                x=[vectors_2d[0, 0]],
                y=[vectors_2d[0, 1]],
                mode='markers+text',
                marker=dict(size=15, color='red'),
                text=central_word,
                name=central_word,
                textposition='middle center',
                textfont=dict(size=12, color='white')
            )
        )

        # Agrega nodos conectados
        fig_connections.add_trace(
            go.Scatter(
                x=vectors_2d[1:, 0],
                y=vectors_2d[1:, 1],
                mode='markers+text',
                marker=dict(size=10),
                text=similar_words,
                name=f'Relacionadas con {central_word}',
                textposition='top center'
            )
        )

        # Agrega líneas de conexión
        for i in range(1, len(vectors_2d)):
            fig_connections.add_shape(
                type="line",
                x0=vectors_2d[0, 0],
                y0=vectors_2d[0, 1],
                x1=vectors_2d[i, 0],
                y1=vectors_2d[i, 1],
                line=dict(color="lightgrey", width=1)
            )

# Layout para el gráfico de conexiones
fig_connections.update_layout(
    title='Conexiones semánticas entre palabras clave (spaCy) en canciones de The Beatles',
    height=600,
    width=800,
    template='plotly_white',
    showlegend=True
)

# el grafico
fig_connections.show()
```

- Se evidencia múltiples instancias de "love" (rojo), "girl" (violeta) y "day" (cian) dispersas en el gráfico.
- Se confirma que el amor, las relaciones personales (girl, boy) y el tiempo son los pilares semánticos de sus letras.


# Comparación NLTK y spaCy

```python
# 8. Comparación con NLTK
try:
    # Carga el modelo NLTK
    nltk_model = Word2Vec.load("beatles_word2vec.model")

    print("\nComparando resultados entre spaCy y NLTK:")

    # Guarda el modelo spaCy para referencia
    model.save("beatles_spacy_word2vec.model")
    print("Modelo spaCy guardado como 'beatles_spacy_word2vec.model'")

    # Compara palabras similares para términos clave
    for palabra in ['love', 'girl']:
        if palabra in model.wv and palabra in nltk_model.wv:
            print(f"\nPalabras similares a '{palabra}':")

            print("Según spaCy:")
            for word, score in model.wv.most_similar(palabra, topn=5):
                print(f"  {word}: {score:.4f}")

            print("\nSegún NLTK:")
            for word, score in nltk_model.wv.most_similar(palabra, topn=5):
                print(f"  {word}: {score:.4f}")

    # Visualización comparativa
    # Crea un DataFrame para visualizar ambos modelos
    common_words = [w for w in model.wv.index_to_key[:100] if w in nltk_model.wv][:50]

    if common_words:
        # Crea un gráfico comparativo
        fig_compare = go.Figure()

        # Agrega puntos para el modelo spaCy
        spacy_vectors = np.array([model.wv[word] for word in common_words])
        spacy_tsne = TSNE(n_components=2, random_state=42, perplexity=min(30, len(common_words)-1))
        spacy_2d = spacy_tsne.fit_transform(spacy_vectors)

        fig_compare.add_trace(
            go.Scatter(
                x=spacy_2d[:, 0],
                y=spacy_2d[:, 1],
                mode='markers+text',
                marker=dict(size=8, color='blue'),
                text=common_words,
                name='spaCy',
                textposition='top center'
            )
        )

        # Agrega puntos para el modelo NLTK
        nltk_vectors = np.array([nltk_model.wv[word] for word in common_words])
        nltk_tsne = TSNE(n_components=2, random_state=42, perplexity=min(30, len(common_words)-1))
        nltk_2d = nltk_tsne.fit_transform(nltk_vectors)

        fig_compare.add_trace(
            go.Scatter(
                x=nltk_2d[:, 0],
                y=nltk_2d[:, 1],
                mode='markers+text',
                marker=dict(size=8, color='red'),
                text=common_words,
                name='NLTK',
                textposition='top center'
            )
        )

        fig_compare.update_layout(
            title='Comparación de embeddings entre spaCy y NLTK',
            height=800,
            width=1200,
            template='plotly_white'
        )

        fig_compare.show()

except Exception as e:
    print(f"No se pudo realizar la comparación con NLTK: {e}")
    # Se guarda el modelo spaCy
    model.save("beatles_spacy_word2vec.model")
    print("Modelo spaCy guardado como 'beatles_spacy_word2vec.model'")

print("\n¡Análisis con spaCy completo! Explora los gráficos para ver las relaciones entre palabras en las canciones de The Beatles.")
```

## Conclusión comparativo
- NLTK se inclina más hacia las co-ocurrencias numéricas y temáticas directas ("money", "lonely"), mientras que spaCy es más amplio en sus asociaciones contextuales, incluyendo referencias a canciones ("Prudence") y conceptos abstractos/sensoriales ("day", "eye"). Las puntuaciones de NLTK son altas que resultan por una posible sobreajuste o un corpus demasiado pequeño/parámetros no óptimos que hacen que las palabras se "peguen" demasiado entre sí.
- NLTK conecta "girl" directamente con los temas centrales de amor, dinero y soledad, y con nombres propios como "pepper". spaCy, por otro lado, ofrece asociaciones más "descriptivas" o "modales" ("little", "heart", "band", "go"), además de la referencia a "Prudence".
- Las puntuaciones de spaCy, aunque también altas, son más "realistas" para la similitud de coseno en Word Embeddings (donde 0.94 ya es muy alto).
- NLTK parece enfocarse más en las co-ocurrencias directas y temáticas fuertes, entre tanto spaCy parece capturar una gama más matizada y contextual de similitude.
- Las similitudes más bajas pero más diversas de spaCy podrían indicar un modelo de embeddings más robusto y generalizable dentro del corpus de los Beatles, que captura una variedad de contextos y no solo las co-ocurrencias más obvias o frecuentes.
- Las similitudes extremadamente altas de NLTK podrían indicar un modelo más sensible a las frecuencias exactas o patrones específicos en el corpus, lo que podría llevar a sobreajuste.
- Se podria sugerir simular con paramétros de entrenamiento mas elevados

- La característica más impactante es la clara separación espacial entre los embeddings generados por spaCy (azules) y NLTK (rojos). Esta separación indica que los dos modelos han aprendido representaciones vectoriales fundamentalmente diferentes para las mismas palabras, incluso si el corpus y el algoritmo (Word2Vec) son los mismos. Sus espacios de significado son distintos.
